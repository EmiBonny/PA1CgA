Questão 1-

```python
num = int(input("Digite um número positivo: "))

if num > 0:
    for i in range(num, 0, -1):
        print(i)
```

Questão 2


```python
num = int(input("Digite um número inteiro positivo: "))

cont = 0

while num > 0:
    cont += 1
    num = num // 10

print("Quantidade de dígitos:", cont)
```

Questão 3


```python
meta = 10000
total = 0
mes = 0

while total < meta:
    deposito = float(input("Valor do depósito: "))
    total = total + deposito
    mes = mes + 1

print("Meses necessários:", mes)
print("Total acumulado:", total)
print("Meta atingida no mês:", mes)
```

Questão 4


```python
while True:
    preco = float(input("Preço do produto: "))

    if preco == 0:
        break

    quantidade = int(input("Quantidade: "))

    subtotal = preco * quantidade

    if subtotal > 1000:
        desconto = 15
    elif subtotal > 500:
        desconto = 10
    elif subtotal > 100:
        desconto = 5
    else:
        desconto = 0

    final = subtotal - (subtotal * desconto / 100)

    print("Subtotal:", subtotal)
    print("Desconto:", desconto, "%")
    print("Valor final:", final)
```
