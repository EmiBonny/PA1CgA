```python
#Exercício 09
produtos = [
    ["Arroz", 25.90],
    ["Feijão", 8.50],
    ["Macarrão", 6.90],
    ["Leite", 5.50],
    ["Café", 15.00]
]

print("Produtos")
for nome, preco in produtos:
    print(f"{nome}: R$ {preco:.2f}")
total = sum(preco for nome, preco in produtos)
print(f"\nTotal: R$ {total:.2f}")

mais_caro = max(produtos, key=lambda item: item[1])
mais_barato = min(produtos, key=lambda item: item[1])

print(f"Mais caro: {mais_caro[0]} (R$ {mais_caro[1]:.2f})")
print(f"Mais barato: {mais_barato[0]} (R$ {mais_barato[1]:.2f})")
```

```python
#crie um programa que permita cadastrar as notas de vários alunos em uma lista. o programa deve:
#1. Perguntar quantos alunos serão cadastrados.
#2. para cada aluno, pedir nome, nota
#3. armazenar as informações em uma lista.
#4.ao final mostrar:
#todos os alunos e suas notas, a media da turma, o aluno com a maior nota, o aluno com a menor nota, quais alunos foram aprovados (nota >=7)
#quais alunos foram reprovados (nota <=7).

qtd = int(input("Qual a quantidade de alunos? "))

nomes = []
notas = []

for i in range(qtd):
    nomes.append(input(f"Nome do aluno {i+1}: "))
    notas.append(float(input(f"Nota {i+1}: ")))

for i in range(qtd):
    print(f"{nomes[i]}: {notas[i]}")

media = sum(notas) / qtd
print(f"Média da turma: {media:.1f}")
maior = notas.index(max(notas))
menor = notas.index(min(notas))
print(f"Maior nota: {nomes[maior]} ({max(notas)})")
print(f"Menor nota: {nomes[menor]} ({min(notas)})")

print("Aprovados:")
for i in range(qtd):
    if notas[i] >= 7:
        print(f"- {nomes[i]}")

print("Reprovados:")
for i in range(qtd):
    if notas[i] < 7:
        print(f"- {nomes[i]}")

```
